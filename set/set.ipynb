```python
# 1. Set banao (duplicate apne aap hat jaate hain)
s = {1, 2, 3, 3, 2}
print(s)
```

```python
# print element one by one
s = {"a", "b", "c"}
for item in s:
    print(item)
```

```python
# add a element in set
s = {1, 2, 3}
s.add(4)
print(s)
```

```python
 # check the element 
s = {10, 20, 30}
i = int(input("enter the number: "))
if i in s:
    print(i, "present in set")
else:
    print(i, "absent in set")
```

```python
# union and intersection
a = {1, 2, 3}
b = {3, 4, 5}
print("Union =", a | b)
print("Intersection =", a & b)
```
