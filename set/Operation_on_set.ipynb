```python
#Add element in a set
my_set = {"A", "B", "C", "D"}

#You can only pass one argument in add attribute
my_set.add("E")             #element add the different location not at end or start
my_set.add("F")
my_set.add("G")
my_set.add("H")
my_set.add("I")
my_set.add("J")
my_set.add("K")
my_set.add("L")

print(my_set)
```

```python
# update (it is work like same as add but the can take mutliple value at a time)
my_set = {"A", "B", "C", "D"}

my_set.update("E", "F", "G", "H","I", "J","K", "L")

print(my_set)
```

```python
# Remove Items from set

my_set = {'H', 'G', 'D', 'F', 'J', 'A', 'I', 'C', 'L', 'E', 'K', 'B'}

my_set.remove("H")
my_set.remove("A")
#my_set.remove("M")          # this line give error because the argument does not exist in set
my_set.discard("M")            #but discard doesn't through if value exist or not in set


print(my_set)
```

```python
mySet = {"apple", "banana", "mango"}
print(mySet)

mySet.add("kiwi")
print(mySet)
mySet.update(["orange", "grapes"])
print(mySet)

mySet.remove("banana")
print(mySet)
mySet.discard("banana") # No error if not present
print(mySet)
mySet.pop() 
print(mySet)
mySet.clear()
```

```python
set1 = {1,2,3}
set2 = {3,4,5}

set3 = set1.union(set2)
print("Union:", set3)

set4 = set1.intersection(set2)
print("Intersection:", set4)

set5 = set1.difference(set2)
print("Difference:", set5)

set6 = set1.symmetric_difference(set2)
print("Symmetric Difference:", set6)

newSet = set1.copy()
print("Copy:", newSet)

set1.update(set2)
print("Updated Set1:", set1)
```
