```python
# Creating a List
fruits = ["apple", "banana", "mango"]
print(fruits)
```

```python
# print all element one by one using loop
numbers = [10, 20, 30, 40, 50]
for n in numbers:
    print(n)
```

```python
# Print a sum of list
numbers = [10, 20, 30, 40, 50]
total = 0
for n in numbers:
    total = total + n
print("Sum =", total)
```

```python
# 4.Find a big number in list
numbers = [45, 12, 89, 33, 7]
big = numbers[0]
for n in numbers:
    if n > big:
        big = n
print("Big Number =", big)
```

```python
# Print even number in list
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
for n in numbers:
    if n % 2 == 0:
        print(n)
```

```python
# Nested list
data = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(data)
```
