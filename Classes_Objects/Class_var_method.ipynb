# Class Variable and Class Method

```python
class clss_var:
    college_name = "SSBT's"    # ==> class variable


clss_var.college_name

clss_var.college_name = "North Maharashtra University"

clss_var.college_name
```

```python
class banking:
    interest_rate = 0.08 

    @classmethod
    def int_rate_calc(cls):
        cls.interest_rate = cls.interest_rate + (cls.interest_rate * 0.02)
        print(cls.interest_rate)
    

banking.interest_rate
banking.int_rate_calc()
```

```python
class Banking:
    Int_Rate = 0.08     # 8%Class Variable

    @classmethod
    def change_int_Rage(cls):
        cls.Int_Rate = cls.Int_Rate + (cls.Int_Rate*0.02)
        print(cls.Int_Rate)

Banking.Int_Rate
Banking.change_int_Rage()
```
