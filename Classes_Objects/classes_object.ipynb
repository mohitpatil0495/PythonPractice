# Classes And Objects


```python
class test:
    def __init__(self, i, j):
        self.i = i
        self.j = j
        
    def multiplication_table(self, i, j):
        
        for x in self.i:
            x = self.i * j
            print(x)
            self.j = self.j + 1
               
    
    def numbertable(self,i):
        for y in range(0, self.i +1):
            print(y)

objtest = test(i=10, j=1)
print("number table")
objtest.numbertable(10)


objtest.multiplication_table()
```
