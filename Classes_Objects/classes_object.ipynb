# <h><b>Static Method<i><b><h>
  

```python
class Decorat:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    @staticmethod
    def hello():                # this decorator with no argument required
        print("Hello")

testing = Decorat("Mohit", 95)

testing.hello()
```
