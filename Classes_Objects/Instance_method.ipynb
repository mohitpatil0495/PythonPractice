# INSTANCE METHOD

```python
class dog:

    def __init__(self, name, breed):
        self.name = name        #instance variable 
        self.breed = breed

    def prnt(self):
        print("Hello", self.name)
        print(self.name, "is a ", self.breed, "dog.")
        


dog = dog("Kalu", "Gavathi")

dog.prnt()
```
