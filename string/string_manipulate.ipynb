```python
# case changing of string
txt = "Hello world"

print(txt)

txt.upper()   # return the whole string in Capital
print(txt)

txt.lower() # return the whole string in lower case
print(txt)

txt.capitalize() # return the first character in Capital
print(txt)
```

```python
#replace 
like = "I like Bananas"

like.replace("I", "We")
print(like)

like.replace("Bananas", "Apples")
print(like)
```

```python
#split 
splt = "I am learning Python Programming"

splt.split()            #This is default always added a space for separate the string
print(splt)

t = splt.split(" ")
print(t)
print(splt)
```

```python
#Extracting string 
name = "rajesh40"

new_name = ""

for i in name:
    if i.isalpha() == True:
        new_name = new_name + i

print(new_name)

num = ""

for i in name:
    if i.isalpha() == False:
        num = num + i

print(num)
```
