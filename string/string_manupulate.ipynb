```python
txt = "Hello world"

print(txt)

txt.upper()   # return the whole string in Capital
print(txt)

txt.lower() # return the whole string in lower case
print(txt)

txt.capitalize() # return the first character in Capital
print(txt)
```

```python
#Extracting string 
name = "rajesh40"

new_name = ""

for i in name:
    if i.isalpha() == True:
        new_name = new_name + i

print(new_name)

num = ""

for i in name:
    if i.isalpha() == False:
        num = num + i

print(num)
```
