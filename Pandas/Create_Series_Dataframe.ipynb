# <b> SERIES <b>

```python
import pandas as pd
```

```python
Data_01 = [1,2,3,4,5]
seres = pd.Series(Data_01)
seres
```

```python
names= ['mohit', 'om', 'sachin', 'chinmay']

srs = pd.Series(names)
print(srs)
```

```python
weekdays = ['sunday', 'monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saterday']

weekdays_series = pd.Series(weekdays)

print(weekdays_series)
```

```python
# Create the Series uisng Dictinary Pandas
calories = {"day1": 420, "day2": 380, "day3": 390}
calSeries = pd.Series(calories)
calSeries
```

```python
# Using Dict, you will get index as well of your choice
Data_01 = {"KEY_1":1,"KEY_2":2,"KEY_3":3}
Data_01
pd.Series(Data_01)
```
