```python
# Creating a Tuple
colors_tup = ("red", "green", "blue")
print(colors_tup)

```

```python
# access the element using the index
colors = ("red", "green", "blue")
#           0       1       2
print(colors[0])
print(colors[2])
```

```python
# 3. Tuple ke elements loop se print karo
nums = (1, 2, 3, 4, 5)
for n in nums:
    print(n)
```

```python
# length and sum of tuple
nums = (10, 20, 30)
total = 0
for n in nums:
    total = total + n
print("Length =", len(nums))
print("Sum =", total)
```

```python
# Checking a particular value present or not in tuple
nums = (5, 10, 15, 20)
i = int(input("Enter the number: "))
if i in nums:
    print(i, "is present")
else:
    print(i, "is not present")
```

```python
# Nested Tuple
data_tup = ((1, 2), (3, 4), (5, 6))
print(data_tup)
```

```python
# print a sum all element in tNested tuple
data_tup = ((1, 2), (3, 4), (5, 6))
total = 0
for t in data_tup:
    for item in t:
        total = total + item
print("Sum =", total)
```
