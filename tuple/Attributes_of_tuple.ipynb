```python
# index
tuple_01 = (5, 4, 5, 2, 5, 3, 5, 4, 1, 5)

print(tuple_01[0])
print(tuple_01[1])
print(tuple_01[2])
print(tuple_01[3])
print(tuple_01[4])
print(tuple_01[5])
print(tuple_01[6])
print(tuple_01[7])
print(tuple_01[8])
print(tuple_01[9])
```

```python
# slicing on tuple
tuple_01 = (5, 4, 5, 2, 5, 3, 5, 4, 1, 5)

print(tuple_01[0:])         # print entire tuple 
print(tuple_01[0:2])

print(tuple_01[::-1])          # print a list in reverse order


#Count how many times a specific value appears in the list

print(tuple_01.count(5))
print(tuple_01.count(2))
print(tuple_01.count(4))
```

```python
# add and join
summer = ("march", "april", "may", "june")
monsoon = ("july", "august", "september", "october")
winter = ("november", "december", "january", "february")

Year = summer + monsoon + winter
print(Year)
```

```python
# repeat values
repeat = ("A", "B", "C", "D") * 5

print(repeat)
```

```python
# item present or not in tuple
months = ('march', 'april', 'may', 'june', 'july', 'august', 'september', 'october', 'november', 'december', 'january', 'february')

print("march" in months)
print("sunday" in months)
print("jan" in months)
print("january" in months)
```

```python

names = ( "Mohit", "Priya", "Vikram", "Shradhha", "Rohan", "Aditi")

status_of_student = input("enter student name: ")

if status_of_student in names:
    print(status_of_student, "is Present")
else:
    print(status_of_student, "is absent")
```

```python
# length 
names = ( "Mohit", "Priya", "Vikram", "Shradhha", "Rohan", "Aditi")
months = ('march', 'april', 'may', 'june', 'july', 'august', 'september', 'october', 'november', 'december', 'january', 'february')
summer = ("march", "april", "may", "june")
monsoon = ("july", "august", "september", "october")
winter = ("november", "december", "january", "february")

print(len(names))
print(len(months))
print(len(summer))
print(len(monsoon))
print(len(winter))

```
