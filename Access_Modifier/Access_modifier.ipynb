```python
# Within the class

class Student:
    def display(self):
        print("Public Method")
        
    def _display_protected(self):
        print("Protected Method")
    def __display_private(self):
        print("Private")
    
    def show(self):
        self.display()
        self._display_protected()
        self.__display_private()

obj = Student()
obj.show()
```
