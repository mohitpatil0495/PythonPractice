# Return and Non Return

```python
# Non Return type Function

def fun(a, b):
    c = a + b
    print(c)



sum = fun(5, 5)

print(sum)
```

```python
def add_values(a,b):
    c = a+b
    return c # return excute's the code(store only single value)

value = add_values(3,2)
value 
```

```python
def sum_values(c,d):
    e = c+d
    return e

newValue = sum_values(value,5)
newValue
```

```python
# Return type Function

def num_table(a):
    
    for i in range(a+1):
        print(i)
    return a
    

one_to_ten = num_table(10)

print(one_to_ten)
```

```python
#yeild is generative keyword

def yld(a):
    for i in range(1, a):
        yield i



print(next(yld(10)))
```

```python
def sum_values():
    pass    # Empty Function

def add_values(a):
    for i in range(1,a):
        sum = i+i
        yield sum

value = add_values(5)

print(next(value))
print(next(value))
print(next(value))
print(next(value))
```
