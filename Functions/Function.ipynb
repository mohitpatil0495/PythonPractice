```python
# Convert list into Dictionary and a list value along their square


list_02 = [1, 2, 3]


def square_list(list_02):
    newdict = {x: x*x for x in list_02}
    return newdict

square_list(list_02)
```

```python
# create new list using function and new list having numbers < 5
list09 = [12,2,10,11,5,3, 5, 4,1,-7, -101, 0]

def sort_list(list09):
    newlist = []
    for i in list09:
        if i < 5:
            newlist.append(i)
            newlist.sort()
    return newlist

sort_list(list09)
```
