```python
def my_func_deco(func):
    def wrapper(x, y):
        if x<y:
            x,y = y,x
            print("Values Swapped")
        func(x,y)
        
    return wrapper
        

@my_func_deco 
def origin(x, y):
    print(x/y)


print(origin(10,5))
print(origin(5,10))
```

```python
# Convert values into int from string

def decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        elif isinstance(y, str):
            y = int(y)
        func(x,y)
    return wrapper



@decor
def devision_1(x,y):
    print(x/y)

devision_1(4, "4")
```

```python
# Convert values into int from string
def my_devi_decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        if isinstance(y, str):
            y = int(y)
            func(x,y)
        if y>x:
            x,y = y,x
            print("Value Swapped")
        return func(x,y)
    return wrapper

@my_devi_decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun('5',10)
```
