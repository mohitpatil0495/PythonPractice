```python
def my_func_deco(func):
    def wrapper(x, y):
        if x<y:
            x,y = y,x
            print("Values Swapped")
        func(x,y)
        
    return wrapper
        

@my_func_deco 
def origin(x, y):
    print(x/y)


print(origin(10,5))
print(origin(5,10))
```

```python
# Convert values into int from string

def decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        elif isinstance(y, str):
            y = int(y)
        func(x,y)
    return wrapper



@decor
def devision_1(x,y):
    print(x/y)

devision_1(4, "4")
```

```python
# Convert values into int from string
def my_devi_decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        if isinstance(y, str):
            y = int(y)
            func(x,y)
        if y>x:
            x,y = y,x
            print("Value Swapped")
        return func(x,y)
    return wrapper

@my_devi_decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun('5',10)
```

```python
# Decorator function to convert strings to integers and square the values
def square_elements(func):
    def wrapper(keys_list, values_list):
        # Convert strings to int and square them for both lists
        squared_keys = [int(x)**2 for x in keys_list]
        squared_values = [int(x)**2 for x in values_list]
        
        # Pass the processed lists to the main function
        return func(squared_keys, squared_values)
    return wrapper

# Main function to create the dictionary
@square_elements
def create_dictionary(keys, values):
    # Combine the two lists into a dictionary using zip
    return dict(zip(keys, values))

# Input lists
List_01 = ['1', '2', '3', '4']
List_02 = ['10', '20', '30', '40']

# Execute the decorated function
result_dict = create_dictionary(List_01, List_02)

print(result_dict)
```
