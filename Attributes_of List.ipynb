```python
 # slicing on list

my_list = ["Apple", "Banana", "Kiwi", "Mango"]
#               0       1       2       3
#               -4      -3      -2      -1
print(my_list[0])
print(my_list[1])
print(my_list[-1])
print(my_list[-4])
```

```python
# append and insert
my_list = ["Apple", "Banana", "Kiwi", "Mango"]

my_list.append("Pineapple")   #element add at the end of list 
print(my_list)

my_list.insert(0, "Orange")    #element add the pass (0) index
print(my_list)

my_list.extend(["Grapes", "Chicu"])
print(my_list)
```

```python
#remove, pop and delete
my_list = ["Apple", "Banana", "Kiwi", "Mango"]

my_list.remove("Apple")    # remove specific item
print(my_list)

my_list.pop(1)          # remove by index
print(my_list)

del my_list[3]          # delete by index
print(my_list)

del my_list             #delete the list 
```

```python
# creating a duplicate of list
l1 = [1, 2, 4, 5]

l2 = l1     # copy the list but have a same memory location

l3 = l1.copy()      # copy the list having different memory location

l3 = l1.copy()

print(l3)
print(id(l3))
print(id(l1))
print(id(l2))
```
