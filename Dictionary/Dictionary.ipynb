```python
# 1. Dictionary banao aur print karo
student = {"name": "Mohit", "course": "Data Science"}
print(student)
```

```python
#print value 
student = {"name": "Mohit", "course": "Data Science"}
print(student["name"])
print(student["course"])
```

```python
# 3. Value update
student = {"name": "Mohit", "course": "Data Science"}
student["course"] = "BCA"
student["city"] = "Jalgaon"
print(student)
```

```python
# print keys and values
marks = {"Python": 85, "ML": 78, "SQL": 92}
for key in marks:
    print(key, ":", marks[key])
```

```python
# 5. total marks
marks = {"Python": 85, "ML": 78, "SQL": 92}
total = 0
for key in marks:
    total = total + marks[key]
print("Total =", total)
```

```python
# Nested Dictionary
students = {
    "s1": {"name": "mohit", "marks": 80},
    "s2": {"name": "om", "marks": 90}
}
print(students)
```

```python
# print nested dictionary using loop
students = {
    "s1": {"name": "mohit", "marks": 80},
    "s2": {"name": "om", "marks": 90}
}
for key in students:
    print(key)
    for k in students[key]:
        print("  ", k, ":", students[key][k])
```
