```python
print("ATTRIBUTES OF DICTIONARY")
```

```python
myDict = {"name": "John", "age": 25, "city": "Pune"}

for i, j in myDict.items():
    print(j, i)
```

```python
myDict = {"name": "John", "age": 25, "city": "Pune"}

for i in myDict.values():
    print(i)
```

```python
myDict = {"name": "John", "age": 25, "city": "Pune"}

for i in myDict.keys():
    print(i)
```

```python
for i in myDict.keys():
    print(myDict.get(i))
```

```python
# Remove items
myDict = {"name": "John", "age": 25, "city": "Pune"}

myDict.pop("age")
print(myDict)
```

```python
myDict = {"name": "John", "age": 25, "city": "Pune"}

myDict.popitem()        # delete the last key pair
print(myDict)
```

```python
# delete key 
myDict = {"name": "John", "age": 25, "city": "Pune"}

del myDict["city"]

print(myDict)
```

```python
# update
myDict = {"name": "John", "age": 25, "city": "Pune"}

new_dict = {"pincode": 425502, "area": "Satpuda"}   

myDict.update({"pincode": 425502, "area": "Satpuda"})

print(myDict)

```

```python
# empty the dictionary

newdict = {'name': 'John', 'age': 25, 'city': 'Pune', 'pincode': 425502, 'area': 'Satpuda'}

newdict.clear()

print(newdict)
print(type(newdict))
```

```python
# merging 
myDict = {"name": "John", "age": 25, "city": "Pune"}

new_dict = {"pincode": 425502, "area": "Satpuda"}   

merged_dict = {**myDict, **new_dict}

print(merged_dict)
```

```python
# join the dictionaries
myDict = {"name": "John", "age": 25, "city": "Pune"}

new_dict = {"pincode": 425502, "area": "Satpuda"}   

myDict.update(new_dict)
``
print(myDict)
```

```python
#sorted
myDict = {"name": "John", "age": 25, "city": "Pune"}

new_dict = {"pincode": 425502, "area": "Satpuda"} 

sorted_dict = dict(sorted(myDict.items()))

print(myDict)

```

```python
# comprehesive
list1 = [1,2,3,4,5]
n_dict = {}

for i in list1:
    n_dict[i] =  i**2

print(n_dict)
```

```python
#comprehensive
list_01 = [1, 2, 3]


{i : i*i for i in list_01}
```

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}
my_dict2 = {"area": "San francisco"}

new_dict = {**my_dict, **my_dict2}
print(new_dict)
new_dict2 = dict(my_dict, **my_dict2)
print(new_dict2)

my_dict.setdefault("country", "India")

print(my_dict)
```

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}

new_dict = dict(reversed(list(my_dict.items())))
print(new_dict)

sort_dict = dict(sorted(my_dict.items()))
print(sort_dict)
```

```python
list = [1,2,3]
newDict = {}

for i in list:
    newDict[i] = i**2

print(newDict)

# Comprehension
# {key : value iterations}
print("Dictionaries Comprehension:", {i : i**2 for i in list})
# {i : i*2 for i in list}
```
