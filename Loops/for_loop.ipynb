```python
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]
col_03 = ["Black","LightGreen","White" ,"Pink"]



for i in col_01:
    for j in col_02:
        for k in col_03:
            if i == j == k:
                print("matched: ", i, j, k )
```
