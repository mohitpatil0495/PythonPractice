```python
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]
col_03 = ["Black","LightGreen","White" ,"Pink"]


for i in (col_01):
    for j in (col_02):
        for k in (col_03):
            if i == j == k:
                idx_1 = col_01.index(i)
                idx_2 = col_02.index(j)
                idx_3 = col_03.index(k)
                print("Matched: ", i, j, k )
                print("Index in col_01: ",idx_1)
                print("Index in col_02: ",idx_2)
                print("Index in col_03: ",idx_3)
                
            
                
```
