```python
txt = "Hello world"

print(txt)

txt.upper()   # return the whole string in Capital
print(txt)

txt.lower() # return the whole string in lower case
print(txt)

txt.capitalize() # return the first character in Capital
print(txt)
```

```python
like = "I like Bananas"

like.replace("I", "We")
print(like)

like.replace("Bananas", "Apples")
print(like)
```
