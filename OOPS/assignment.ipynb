# <i><b> Single Inheritance<i><b>

```python
class BasicTV:
    def change_channel(self):
        print("Channel change ho gaya.")

class SmartTV(BasicTV):  # SmartTV ne BasicTV ko inherit kiya
    def watch_netflix(self):
        print("Netflix chal raha hai.")

# Object banate hain
my_tv = SmartTV()
my_tv.watch_netflix()    # Naya feature
my_tv.change_channel()   # Parent (BasicTV) ka feature
```

```python
class FoodOrder:
    def __init__(self, order_id, restaurant, price):
        self.order_id = order_id
        self.restaurant = restaurant
        self.price = price

    def show_order(self):
        print(f"Order #{self.order_id} from {self.restaurant}: Rs.{self.price}")


class PizzaOrder(FoodOrder):
    def __init__(self, order_id, restaurant, price, extra_cheese):
        super().__init__(order_id, restaurant, price)
        self.extra_cheese = extra_cheese

    def prepare(self):
        cheese = "with extra cheese" if self.extra_cheese else "regular cheese"
        print(f"Baking pizza {cheese}...")


order = PizzaOrder(101, "Pizza Hub", 349, True)
order.show_order()   # inherited
order.prepare()      # own method
```

# <i><b>Multiple Inheritance <i><b>

```python
class Watch:
    def show_time(self):
        print("2:00 AM")

class FitnessTracker:
    def count_steps(self):
        print("5000 steps.")

class SmartWatch(Watch, FitnessTracker):  
    def receive_notification(self):
        print("WhatsApp message!")


my_smartwatch = SmartWatch()
my_smartwatch.show_time()               # Watch class se aaya
my_smartwatch.count_steps()             # FitnessTracker class se aaya
my_smartwatch.receive_notification()    # Apna khud ka feature
```

```python
class User:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def login(self):
        print(f"{self.name} logged in.")


class Customer(User):
    def __init__(self, name, phone, address):
        super().__init__(name, phone)
        self.address = address

    def place_order(self, item):
        print(f"{self.name} ordered {item} to {self.address}.")


class PremiumCustomer(Customer):
    def __init__(self, name, phone, address, membership="Gold"):
        super().__init__(name, phone, address)
        self.membership = membership

    def get_free_delivery(self):
        print(f"{self.membership} member {self.name} gets FREE delivery!")


p = PremiumCustomer("Mohit", "90XXXXXX19", "Jalgaon", "Gold")
p.login()
p.place_order("Paneer Tikka")
p.get_free_delivery()
```

# <i><b> Multilevel Inheritance <i><b>

```python
"""Multilevel inheritance ek chain ki tarah kaam karta hai. 
Grandparent se Parent mein properties aati hain, aur Parent se Child mein. 
Tech ki duniya mein dekhein toh sabse pehle aaya Computer jiska kaam tha data process karna. 
Phir aaya Laptop, jisne Computer ke saare features inherit kiye aur apna naya feature 
(Battery/Portability) add kar diya. Aur aaj ke time mein aaya GamingLaptop, 
jisne normal Laptop ke features inherit kiye aur usme 'High-End Graphics' daal diye. 
Toh Gaming Laptop ke paas teeno ki taakat hai!"""
```

```python
class Computer:
    def process_data(self):
        print("Data process ho raha hai.")

class Laptop(Computer):  # Laptop ne Computer ko inherit kiya
    def run_on_battery(self):
        print("Battery par chal raha hai.")

class GamingLaptop(Laptop):  # GamingLaptop ne Laptop ko inherit kiya
    def play_heavy_games(self):
        print("GTA V chal raha hai bina lag ke!")

# Object banate hain sabse latest child ka
my_setup = GamingLaptop()
my_setup.play_heavy_games()   # Apna khud ka feature
my_setup.run_on_battery()     # Laptop class se aaya
my_setup.process_data()       # Sabse upar Computer class se aaya
```
