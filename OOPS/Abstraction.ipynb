# Abstraction

Abstraction means hiding the complex implementation and showing only the important details to the user.

For example, when we drive a car, we only care about starting the engine, accelerating, and braking. We do not need to know how the engine works internally.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        return "Car started"

    def stop(self):
        return "Car stopped"

class Bike(Vehicle):
    def start(self):
        return "Bike started"

    def stop(self):
        return "Bike stopped"

car = Car()
bike = Bike()

print(car.start())
print(bike.stop())
```

### Real-world idea

In real life, a remote control is a good example of abstraction. We press a button to switch channels, but we do not see the internal electronic logic.

Similarly, in Python, abstraction hides unnecessary details and exposes only the important actions.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        return f"Paid {amount} using Credit Card"

class UpiPayment(Payment):
    def pay(self, amount):
        return f"Paid {amount} using UPI"

payments = [CreditCardPayment(), UpiPayment()]

for method in payments:
    print(method.pay(500))
```
