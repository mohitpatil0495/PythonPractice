```python
# simple method without overloading

class mycls:
    def sum_4(self, a, b, c, d):
        return a+b+c+d
    def sum_3(self, a, b, c, ):
        return a+b+c
    def sum_2(self, a, b):
        return a+b
obj = mycls()
print(obj.sum_4(2,4,5,-1))
print(obj.sum_3(1, 2, 3))
obj.sum_2(1,9)
```

```python
class mohit:

    def sum(self, a=None, b=None, c=None, d=None):
        if a != None and b != None and c != None and d!=None:
            s = a+b+c
        elif a != None and b != None and c != None:
            s = a+b+c
        elif a!=None and b!=None:
            s = a+b
        else:
            print("Provide atleast two numbers ")
        
        return s

        
my_obj = mohit()
my_obj.sum(10,20,30)

```

```python
# overridng

class Animal:
    def sound(self):
        return "Animals makes sound"

class Dog:
    def sound(self):
        return "barks "
    
class Cat(Animal):
    def sound(self):
        return "meow meow"
    
cat = Cat()
cat.sound()
```

```python
class Interest:
    def Interest(self):
        return 0.11
    
class Home_interest(Interest):
    def Interest(self):
        return 0.08


hi = Home_interest()
hi.Interest()
```
