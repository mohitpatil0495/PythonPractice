# <i> <b>Single Inheritance <b><i>

```python
class Parent:
    def parent_class(self):
        print("This is Parent Class")
    
class Child(Parent):
    def child_class(self):
        return f"This Child Class"
    
P_obj = Parent()
P_obj.parent_class()
C_obj = Child()

C_obj.child_class()

```

```python
class Car:
    @staticmethod
    def start():
        print("Car started...")
    @staticmethod
    def stop():
        print("Car Stopped... ")
    
class ToyotaCar(Car):
    def __init__(self, name):
        self.name = name
    
t_obj = ToyotaCar("Fortunure")

t_obj.name
```

# Multilevel Inheritance

```python
class Gradpa:
    def Gradpa_method(self):
        return "I have 28 acres "

class father(Gradpa):
    def father_method(self):
        return "I have 5 acres "
    
class childd(father, Gradpa):
    def child_method(self):
         return "I have Books"

G_obj = Gradpa()
G_obj.Gradpa_method()

P_obj = father()
P_obj.father_method()
Cc_obj = Childd()
Cc_obj.Gradpa




```
