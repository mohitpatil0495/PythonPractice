```python
#nested loop 
#find matching element from two lists

list_1 = ["red", "yellow", "lightgreen"]
list_2 = ["white", "black", "yellow"]

for i in list_1:
    for j in list_2:
        if i == j:
            print("matched: ", i, j)
```

```python
list_1 = ["red", "yellow", "lightgreen"]
list_2 = ["white", "black", "yellow"]
list_3 = [ "white", "purple", "yellow"]

for i in list_1:
    for j in list_2:
        for k in list_3:
            if i == j == k:
                print ("matched: ", i, j, k)
```

```python
list_1 = ["red", "yellow", "lightgreen","blue" ]
list_2 = ["white", "black","blue", "yellow"]
list_3 = [ "white", "blue","purple", "yellow"]

for i in list_1:
    for j in list_2:
        for k in list_3:
            if i == j == k:
                print ("matched: ", i, j, k)

```
