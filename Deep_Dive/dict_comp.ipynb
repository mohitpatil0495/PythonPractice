## More dictionary comprehension examples

The general form is `{key_expression: value_expression for item in iterable if condition}`. The `if` condition is optional.

```python
# Map each number to its square
squares = {number: number ** 2 for number in range(1, 6)}
print(squares)
```

```python
# Keep only even numbers and map each one to its square
even_squares = {number: number ** 2 for number in range(1, 11) if number % 2 == 0}
print(even_squares)
```

```python
# Use strings as keys and their lengths as values
words = ["python", "dictionary", "loop"]
word_lengths = {word: len(word) for word in words}
print(word_lengths)
```

```python
# Choose each value with a conditional expression
number_types = {number: "even" if number % 2 == 0 else "odd" for number in range(1, 6)}
print(number_types)
```

```python
# Transform values from an existing dictionary using .items()
prices = {"book": 12.0, "pen": 1.5, "bag": 25.0}
prices_with_tax = {item: round(price * 1.1, 2) for item, price in prices.items()}
print(prices_with_tax)
```
