# <B> Enumerate <B>

```python
lis =  [1,3,4,5,6,8,0]

for i in enumerate(lis):
    print(i)
```

```python
fruits = ["apple", "banana", "mango"]

for index, Value in enumerate(fruits):
    print(index)
    print(Value)
```

```python
fruits = ["apple", "banana", "mango"]

enumrate = enumerate(fruits)

print(enumrate)     #===> Iterator

print(next(enumrate))
print(next(enumrate))
print(next(enumrate))
```
