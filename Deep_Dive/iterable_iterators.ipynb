# Python Iterable

```python
list1 = [1,2,3,4,5,6,7,8]
def extractonlyevennumber(i):
    return i%2==0

abc = filter(extractonlyevennumber, list1)      # iterator 
print(abc)
print(dir(abc))

for i in abc:
    print(i)
```
