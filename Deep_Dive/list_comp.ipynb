```python
# Create a list of squares for the numbers from 0 through 9
squares = [number ** 2 for number in range(10)]

# Keep only the even squares
even_squares = [square for square in squares if square % 2 == 0]

print(even_squares)
```

```python
list3 = []

for i in range (0, 15):
    
    if i<5:
        list3.append(i*2)
    elif i<10:
        list3.append(i/2)
    else:
        list3.append(i+2)

print(list3)
```
